```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_orders = df['Order ID'].nunique()
profit_margin = total_profit / total_sales * 100
avg_order_value = total_sales / total_orders

print(f"Total Sales      : {total_sales:,.2f}")
print(f"Total Profit     : {total_profit:,.2f}")
print(f"Total Orders     : {total_orders:,}")
print(f"Profit Margin    : {profit_margin:.2f}%")
print(f"Avg Order Value  : {avg_order_value:,.2f}")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.histplot(df['Sales'], bins=20, ax=axes[0, 0])
axes[0, 0].set_title('Distribution of Sales')

sns.histplot(df['Discount'], bins=15, ax=axes[0, 1])
axes[0, 1].set_title('Distribution of Discount')

sns.histplot(df['Profit'], bins=20, ax=axes[1, 0])
axes[1, 0].set_title('Distribution of Profit')

sns.boxplot(x=df['Profit'], ax=axes[1, 1])
axes[1, 1].set_title('Profit Boxplot (outliers shown as dots)')

plt.tight_layout()
plt.show()
```

catagory

```python
cat = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Orders=('Order ID', 'nunique')
).reset_index()

cat['Profit_Margin_%'] = (cat['Total_Profit'] / cat['Total_Sales'] * 100).round(2)
cat = cat.sort_values('Total_Sales', ascending=False)
print(cat)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.barplot(data=cat, x='Total_Sales', y='Category', ax=axes[0])
axes[0].set_title('Total Sales by Category')
sns.barplot(data=cat, x='Total_Profit', y='Category', ax=axes[1])
axes[1].set_title('Total Profit by Category')
plt.tight_layout()
plt.show()
```

sub catagory


```python
sub = df.groupby('Sub Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
).reset_index()
sub['Profit_Margin_%'] = (sub['Total_Profit'] / sub['Total_Sales'] * 100).round(2)
sub = sub.sort_values('Total_Sales', ascending=False)
print(sub)

plt.figure(figsize=(10, 9))
sns.barplot(data=sub, x='Total_Sales', y='Sub Category')
plt.title('Total Sales by Sub Category')
plt.tight_layout()
plt.show()
```

Yearly sales and profit

```python
df['Year'] = df['Order Date'].dt.year
yearly = df.groupby('Year').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
).reset_index()
yearly['Profit_Margin_%'] = (yearly['Total_Profit'] / yearly['Total_Sales'] * 100).round(2)
print(yearly)

plt.figure(figsize=(8, 5))
sns.barplot(data=yearly, x='Year', y='Total_Sales')
plt.title('Total Sales by Year')
plt.show()
```

Monthly sales trend

```python
monthly = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()
monthly.index = monthly.index.to_timestamp()
print(monthly.head(12))

plt.figure(figsize=(14, 5))
plt.plot(monthly.index, monthly.values, marker='o')
plt.title('Monthly Sales Trend (2015-2018)')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.grid(True)
plt.show()

# Which calendar month has the highest sales overall?
by_month = df.groupby(df['Order Date'].dt.month_name())['Sales'].sum().sort_values(ascending=False)
print(by_month)
```

Regional sales and profit

```python
reg = df.groupby('Region').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Orders=('Order ID', 'nunique')
).reset_index()
reg['Profit_Margin_%'] = (reg['Total_Profit'] / reg['Total_Sales'] * 100).round(2)
reg = reg.sort_values('Total_Sales', ascending=False)
print(reg)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.barplot(data=reg, x='Region', y='Total_Sales', ax=axes[0])
axes[0].set_title('Total Sales by Region')
sns.barplot(data=reg, x='Region', y='Total_Profit', ax=axes[1])
axes[1].set_title('Total Profit by Region')
plt.tight_layout()
plt.show()
```

Discount vs Profit (two-column analysis)

```python
print(df[['Sales', 'Discount', 'Profit']].corr().round(3))

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Discount', y='Profit', alpha=0.3)
plt.title('Discount vs Profit')
plt.show()

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Sales', y='Profit', alpha=0.3)
plt.title('Sales vs Profit')
plt.show()
```
